```python
import polars as pl, numpy as np
d = pl.read_parquet("vectors.parquet")
V = np.array(d["vector"].to_list())
V = V / np.linalg.norm(V, axis=1, keepdims=True)
names, roles = d["player"].to_list(), d["role"].to_list()

def similar(sub, k=8):
    i = next(j for j, n in enumerate(names) if sub.lower() in n.lower())
    print(f"\n~ {names[i]} ({roles[i]})")
    s = V @ V[i]
    for j in np.argsort(-s)[1:k+1]:
        print(f"  {s[j]:.3f}  {names[j]}  ({roles[j]})")

similar("Messi")
similar("Busquets")
similar("Piqué")
```
